```python
# Analyze the raw data values and scaling
print("=== RAW DATA ANALYSIS ===")

X = data['X']
y = data['y']

print(f"Raw data sample (first trial, first channel, first 10 timepoints):")
print(X[0, 0, :10])

print(f"\nData statistics:")
print(f"Min value: {X.min():.2e}")
print(f"Max value: {X.max():.2e}")
print(f"Mean: {X.mean():.2e}")
print(f"Std: {X.std():.2e}")

print(f"\nConverted to microvolts (×1e6):")
print(f"Min: {X.min()*1e6:.2f} μV")
print(f"Max: {X.max()*1e6:.2f} μV")
print(f"Mean: {X.mean()*1e6:.2f} μV")
print(f"Std: {X.std()*1e6:.2f} μV")

print(f"\nEEG context:")
print(f"- Typical EEG range: ±50-200 μV")
print(f"- Your data range: ±{max(abs(X.min()*1e6), abs(X.max()*1e6)):.1f} μV")

if max(abs(X.min()*1e6), abs(X.max()*1e6)) < 100:
    print("✅ Data is in normal EEG range")
else:
    print("⚠️ Data exceeds typical EEG range")
```
